# 🧩 Ejercicio 3 – Regresión Lineal: Temperatura vs Consumo Energético

## 🎯 Objetivo
Explorar si existe una relación lineal entre la **temperatura media mensual** y el **consumo energético** en una región, utilizando **regresión lineal simple**.

Usaremos datos simulados representando el comportamiento mensual durante un año, con tendencia esperada a **mayor consumo cuando la temperatura baja**.

---

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(0)

# Simulamos datos para 12 meses
meses = np.arange(1, 13)
temperaturas = 25 - 10 * np.sin((meses - 1) * np.pi / 6) + np.random.normal(0, 1, 12)  # patrón estacional
consumo = 200 + (25 - temperaturas) * 10 + np.random.normal(0, 5, 12)  # más consumo en meses fríos

df = pd.DataFrame({
    'mes': meses,
    'temperatura_media': temperaturas,
    'consumo_energia': consumo
})

df
```

# 🧩 Ejercicio 4 – Ajuste robusto: Altura vs Edad

## 🎯 Objetivo
Comparar el rendimiento entre un modelo de **regresión lineal estándar** y **modelos robustos** al ajustar la relación entre edad y altura en una población simulada.

Se incluirán algunos **outliers artificiales** (errores o valores atípicos) para evaluar la diferencia entre:

- Regresión Lineal Ordinaria
- Huber Regressor
- RANSAC
- Theil-Sen Estimator

---

Comparar el rendimiento entre un modelo de **regresión lineal estándar** y **modelos robustos** al ajustar la relación entre edad y altura.

En esta versión se generará un conjunto principal de datos normales, y un **subconjunto separado de outliers**, que luego serán combinados para simular mediciones atípicas realistas.

---

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

# Datos normales
n = 60
edad_normal = np.linspace(5, 18, n)
altura_normal = 100 + 5 * edad_normal + np.random.normal(0, 5, n)
df_normal = pd.DataFrame({'edad': edad_normal, 'altura': altura_normal})

# Subconjunto de outliers
n_out = 10
edad_out = np.random.uniform(5, 18, n_out)
altura_out = np.random.uniform(220, 300, n_out)  # alturas imposibles
df_outliers = pd.DataFrame({'edad': edad_out, 'altura': altura_out})

# Combinar
df = pd.concat([df_normal, df_outliers], ignore_index=True)
df.head()
```

Utilizaremos una muestra de galaxias del **SDSS** y su relación entre color y magnitud en banda `r`, emulando un **diagrama color-magnitud**.

---

```python
from astroquery.sdss import SDSS
from astropy.table import Table
from astropy import coordinates as coords
import pandas as pd

query = """
SELECT TOP 3000
    ra, dec, modelMag_g, modelMag_r,
    modelMag_g - modelMag_r AS color_gr
FROM PhotoObj
WHERE type = 3 -- galaxias
AND modelMag_r BETWEEN 14 AND 20
AND modelMag_g BETWEEN 14 AND 22
AND (modelMag_g - modelMag_r) BETWEEN 0 AND 2
"""

# Ejecutar consulta
res = SDSS.query_sql(query)
df = res.to_pandas()
df = df.dropna()
df.head()
```

# 🧩 Ejercicio 11 – Clustering en la Red Sequence del Cúmulo de Coma

## 🎯 Objetivo
Aplicar técnicas de **clustering** (K-Means, DBSCAN) sobre galaxias del **cúmulo de Coma**, en su diagrama color-magnitud, para:

- Identificar la **Red Cluster Sequence**,
- Separar galaxias pasivas y activas,
- Detectar posibles **subestructuras internas**.

---

```python
from astroquery.sdss import SDSS
import pandas as pd

query = '''
SELECT TOP 3000
    modelMag_g, modelMag_r, modelMag_i, modelMag_z,
    modelMag_r - modelMag_i AS color_ri,
    modelMag_g - modelMag_r AS color_gr,
    modelMag_r,
    z AS redshift
FROM PhotoObjAll
WHERE ra BETWEEN 193 AND 195
AND dec BETWEEN 26 AND 28
AND type = 3
AND modelMag_r BETWEEN 14 AND 20
AND z BETWEEN 0.01 AND 0.05
'''

res = SDSS.query_sql(query)
df = res.to_pandas()
df = df.dropna()
df.head()
```
