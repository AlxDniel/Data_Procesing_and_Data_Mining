# 🧠 Sección I: Fundamentos y Motivación del Machine Learning

Esta sección introduce los conceptos esenciales del aprendizaje automático (Machine Learning, ML), su motivación y cómo se aplica a problemas astronómicos.


## 🤖 ¿Qué es Machine Learning?

Machine Learning es el estudio de algoritmos que **aprenden automáticamente a partir de los datos** para hacer predicciones o descubrir patrones sin una programación explícita.

> **Definición de Tom Mitchell (1997):**  
> Un programa de computadora aprende de la experiencia E con respecto a una tarea T y una medida de rendimiento P, si su rendimiento en T, medido por P, mejora con E.

En contraste con los modelos físicos tradicionales, ML:
- Se adapta a relaciones complejas no lineales
- Puede trabajar con grandes volúmenes de datos
- Requiere validación estadística en lugar de derivación teórica

## 🤖 1. ¿Qué es Machine Learning?

ML es el diseño de algoritmos que aprenden automáticamente patrones a partir de los datos, sin estar explícitamente programados para cada tarea.

> "A computer program is said to learn from experience E with respect to some class of tasks T and performance measure P..." — *Tom Mitchell (1997)*

Aplicaciones en astronomía:
- Clasificación de objetos
- Descubrimiento de exoplanetas
- Agrupamiento de galaxias
- Análisis espectroscópico


## 📚 Tipos de aprendizaje

## 📈 4. Conceptos estadísticos clave

### 📊 Medidas estadísticas

- **Media (promedio):**
$ bar{x} = \frac{1}{n} \sum_{i=1}^{n} x_i $

- **Varianza:**
$ \sigma^2 = \frac{1}{n} \sum_{i=1}^{n} (x_i - \bar{x})^2 $

- **Desviación estándar:**
$ \sigma = \sqrt{\sigma^2} $

---

### 🔔 Distribución normal

- Función de densidad:
$ f(x) = \frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{(x - \mu)^2}{2\sigma^2}} $

La mayoría de las variables astronómicas (errores fotométricos, ruido instrumental) se modelan con distribuciones normales.

---

### 🧮 Métricas comunes en ML

- **Error cuadrático medio (MSE):**
$ \text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 $

- **Precisión:**
$ \text{Precisión} = \\frac{\text{TP}}{\text{TP + FP}} $

- **Recall (Sensibilidad):**
$ \text{Recall} = \frac{\text{TP}}{\text{TP + FN}} $

- **F1 Score (media armónica de precisión y recall):**
$ \text{F1} = 2 \cdot \frac{\text{Precisión} \cdot \text{Recall}}{\text{Precisión + Recall}} $

---

### 🔁 Validación cruzada

Método para evaluar modelos usando subconjuntos del conjunto de datos.  
Ejemplo: validación cruzada k-fold divide los datos en *k* partes, entrena con *k-1* y prueba con la restante.

---

### ⚠️ Overfitting vs Underfitting

- **Overfitting:** el modelo se ajusta demasiado a los datos de entrenamiento y no generaliza.
- **Underfitting:** el modelo no captura patrones relevantes ni siquiera en el entrenamiento.

Ambos pueden detectarse mediante curvas de aprendizaje y validación cruzada.

# 📘 Sección II: Introducción al Aprendizaje Supervisado

Esta sección cubre los fundamentos del aprendizaje supervisado, comenzando por la regresión lineal como modelo base.


## 🧭 1. ¿Qué es el Aprendizaje Supervisado?

El aprendizaje supervisado utiliza datos etiquetados: cada muestra de entrada ($ \vec{x} $) tiene una salida ( y ) conocida.

- **Objetivo:** Aprender una función ( $f(\vec{x}) \approx y )$
- **Ejemplos:**
  - Clasificar tipos estelares a partir de colores
  - Estimar masa galáctica desde luminosidad

Tipos:
- **Regresión:** salida continua (ej. luminosidad)
- **Clasificación:** salida categórica (ej. tipo espectral)

## 📈 2. Regresión Lineal Simple

Es el modelo supervisado más básico. Busca una relación lineal entre una variable independiente  (x) y una dependiente (y):

$ y = mx + b $

Donde:
- ( m ) es la pendiente
- ( b ) es la ordenada al origen

**Objetivo:** Minimizar el error cuadrático medio (MSE):

$ \text{MSE} = \frac{1}{n} \sum_{i=1}^n (y_i - \hat{y}_i)^2 $

## 📉 3. Regresión Polinomial

Ajuste de la forma:

$ y = a_0 + a_1 x + a_2 x^2 + \dots + a_d x^d $

**Ventaja:** Mayor flexibilidad para patrones no lineales  
**Riesgo:** Sobreajuste si el grado es muy alto

Se puede usar `PolynomialFeatures` para generar los términos.

````python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

np.random.seed(0)
Teff = np.random.normal(5000, 1000, 100)
Lum = 0.0006 * Teff + np.random.normal(0, 2, 100)

X = Teff.reshape(-1, 1)
y = Lum

model = LinearRegression().fit(X, y)
y_pred = model.predict(X)

# Comparación entre grados 1, 2 y 3
plt.figure(figsize=(8, 5))
x_plot = np.linspace(min(Teff), max(Teff), 200).reshape(-1, 1)

for degree in [1,2, 3, 4,5]:
    model = make_pipeline(PolynomialFeatures(degree), LinearRegression())
    model.fit(X, y)
    y_plot = model.predict(x_plot)
    plt.plot(x_plot, y_plot, label=f"Grado {degree}")

plt.scatter(Teff, Lum, alpha=0.5, label="Datos")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Comparación de ajustes polinomiales")
plt.legend()
plt.show()
````

## ✅ 5. Evaluación del Modelo

### Métricas comunes:
- **MSE (Error Cuadrático Medio)**: penaliza grandes errores
- **R² (Coeficiente de determinación)**: proporción de varianza explicada
- **Visualización de residuos**: diferencias entre valores reales y predichos

### Sobreajuste:
- Alto R² en entrenamiento, pero mal rendimiento en test
- Curvas de aprendizaje ayudan a detectarlo

Usar `train_test_split` y validación cruzada (`cross_val_score`) para evaluación robusta.


# 📘 Sección III: Regresiones Robustas y Clasificación Supervisada

En esta sección abordamos modelos de regresión más resistentes a outliers y métodos clásicos de clasificación.

## 🧱 1. Regresiones robustas

La regresión lineal minimiza el error cuadrático. Esto la hace muy sensible a valores atípicos (outliers).

Los modelos robustos, en cambio, **reducen la influencia de los outliers** en el ajuste.


### 🔧 Modelos robustos

- **HuberRegressor**: combina MSE para valores pequeños y MAE para outliers.
- **RANSAC**: itera múltiples ajustes sobre subconjuntos y elige el mejor basado en consenso.
- **TheilSenRegressor**: ajusta múltiples regresiones usando medianas para mayor robustez.

Usaremos `sklearn.linear_model` para compararlos.

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, HuberRegressor, RANSACRegressor, TheilSenRegressor

np.random.seed(1)
x = np.linspace(0, 10, 100)
y = 2.5 * x + 1 + np.random.normal(0, 5.5, 100)

# Agregar outliers
x_out = np.append(x, [2, 5, 8, 3, 7, 3, 8,9,10])
y_out = np.append(y, [30, -20, 25, -10, 5, -13, 15,-10,-17])
X = x_out.reshape(-1, 1)

models = {
    'Lineal': LinearRegression(),
    'Huber': HuberRegressor(),
    'RANSAC': RANSACRegressor(),
    'TheilSen': TheilSenRegressor()
}

plt.figure(figsize=(10, 6))
for name, model in models.items():
    model.fit(X, y_out)
    y_pred = model.predict(X)
    plt.plot(x, model.predict(x.reshape(-1, 1)), label=name)

plt.scatter(x_out, y_out, c='black', alpha=0.6, label='Datos + outliers')
plt.xlabel("x")
plt.ylabel("y")
plt.title("Comparación de regresiones robustas")
plt.legend()
plt.show()
````

## 🧠 2. Introducción a la clasificación supervisada

Problemas donde la salida es **categórica**

### Modelos comunes:
- **KNN (K vecinos más cercanos):** predice por mayoría en el vecindario
- **Árbol de decisión:** construye reglas jerárquicas

Se evalúan mediante:
- Accuracy (exactitud)
- Precisión y recall
- F1 Score
- Matriz de confusión

````python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import confusion_matrix, classification_report

X, y = make_classification(n_samples=400, n_features=2, n_informative=2,
                           n_redundant=0, n_clusters_per_class=1, random_state=42)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Entrenar modelos
knn = KNeighborsClassifier(n_neighbors=5).fit(X_train, y_train)
tree = DecisionTreeClassifier(max_depth=4).fit(X_train, y_train)

# Evaluar
print("KNN:\n", classification_report(y_test, knn.predict(X_test)))
print("Árbol de decisión:\n", classification_report(y_test, tree.predict(X_test)))

# Visualizar
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
ax[0].scatter(X_test[:,0], X_test[:,1], c=knn.predict(X_test), cmap='coolwarm', alpha=0.7)
ax[0].set_title("KNN")

ax[1].scatter(X_test[:,0], X_test[:,1], c=tree.predict(X_test), cmap='coolwarm', alpha=0.7)
ax[1].set_title("Árbol de decisión")
plt.show()
````

# 📘 Sección IV: Aprendizaje No Supervisado

Esta sección explora técnicas de aprendizaje automático sin etiquetas previas. El modelo intenta descubrir estructuras o patrones por sí solo.


## 🧭 ¿Qué es el aprendizaje no supervisado?

- El modelo **no tiene etiquetas** de clase o variable objetivo.
- El objetivo es **descubrir estructuras** en los datos, como agrupamientos o dimensiones latentes.

Aplicaciones en astronomía:
- Agrupar galaxias por propiedades físicas
- Identificar estructuras en cúmulos
- Encontrar poblaciones estelares similares

## 🔗 Clustering

El objetivo del *clustering* es dividir un conjunto de datos en grupos (clusters) **internamente homogéneos** y **externamente distintos**.

### Modelos:
- **KMeans:** asigna puntos a k centroides. Requiere número de clusters como entrada.
- **DBSCAN:** basado en densidad, identifica regiones densas separadas por zonas vacías.
- **Jerárquico:** agrupa datos en forma de árbol (dendrograma).

## 🧮 Reducción de dimensionalidad: PCA

**PCA (Análisis de Componentes Principales)** busca una representación de los datos en un espacio de menor dimensión conservando la mayor varianza posible.

Usos en astronomía:
- Visualizar grandes catálogos multivariables
- Preprocesamiento para clasificación

````python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
import matplotlib.pyplot as plt

X, _ = make_blobs(n_samples=300, centers=3, cluster_std=0.7, random_state=0)

kmeans = KMeans(n_clusters=3).fit_predict(X)
dbscan = DBSCAN(eps=0.4, min_samples=8).fit_predict(X)
hclust = AgglomerativeClustering(n_clusters=3).fit_predict(X)

fig, ax = plt.subplots(1, 3, figsize=(15, 4))
ax[0].scatter(X[:,0], X[:,1], c=kmeans, cmap='tab10')
ax[0].set_title("KMeans")
ax[1].scatter(X[:,0], X[:,1], c=dbscan, cmap='tab10')
ax[1].set_title("DBSCAN")
ax[2].scatter(X[:,0], X[:,1], c=hclust, cmap='tab10')
ax[2].set_title("Clustering jerárquico")
plt.tight_layout()
plt.show()

# Aplicar PCA a los blobs
X_std = StandardScaler().fit_transform(X)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_std)

plt.figure(figsize=(6,5))
plt.scatter(X_pca[:,0], X_pca[:,1], c=dbscan, cmap='tab10')
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("Visualización con PCA")
plt.grid(True)
plt.show()
````

# 📘 Sección V: Modelos complejos y automatización

En esta sección aprenderemos a estructurar flujos de trabajo en ML para experimentación reproducible, validación y optimización de hiperparámetros.

## 🔄 ¿Por qué automatizar?

Cuando se trabaja con modelos reales:

- Queremos aplicar **múltiples pasos (escalado, PCA, modelo)** en una misma estructura.
- Evitamos **errores humanos** o sesgos en el orden de procesamiento.
- Evaluamos **diferentes configuraciones** (ej. n vecinos, profundidad, etc.) de forma sistemática.

Para esto usamos:
- `Pipeline` → secuencia de pasos automatizados
- `GridSearchCV` → búsqueda de la mejor combinación de hiperparámetros
- `cross_val_score` → evaluación con validación cruzada

````python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import Ridge
from sklearn.model_selection import cross_val_score
import matplotlib.pyplot as plt
from sklearn.model_selection import GridSearchCV


# Datos sintéticos
np.random.seed(42)
x = np.linspace(-20, 20, 100)
y = x**4 - x**5 + 0.5 * x**3 - x + 2 + np.random.normal(-500, 500, 100)

X = x.reshape(-1, 1)

# Pipeline con escalado, polinomio y Ridge
pipe = Pipeline([
    ('scale', StandardScaler()),
    ('poly', PolynomialFeatures(degree=5)),
    ('model', Ridge(alpha=10))
])

#plt.scatter(X,y)
# Validación cruzada
scores = cross_val_score(pipe, X, y, cv=5, scoring='neg_mean_squared_error')

pipe.fit(X, y)
x_plot = np.linspace(min(x), max(x), 200).reshape(-1, 1)
y_plot = pipe.predict(x_plot)

plt.figure(figsize=(8,5))
plt.scatter(x, y, label="Datos", alpha=0.6)
plt.plot(x_plot, y_plot, color='red', label="Mejor ajuste")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
````

| Paso    | Acción                                                              |
| ------- | ------------------------------------------------------------------- |
| `poly`  | Genera características polinómicas (x, x², x³, etc.)                |
| `model` | Aplica regresión Ridge (una forma de regresión lineal regularizada) |


## 🔧 Búsqueda de hiperparámetros

Usamos `GridSearchCV` para buscar combinaciones óptimas de parámetros del modelo (por ejemplo: grado del polinomio, regularización, número de vecinos).

Aquí se entrena el modelo. Lo que hace GridSearchCV es:

    Probar todas las combinaciones de parámetros dados:

        Grados del polinomio: 1, 2, 3

        Valores de regularización (alpha): 0.01, 0.1, 1.0, 10.0

    Para cada combinación:

        Se entrena el pipeline completo con fit() (esto incluye transformar X con PolynomialFeatures, luego aplicar Ridge.fit()).

        Evalúa con validación cruzada (5-fold): divide los datos en 5 partes, entrena en 4 y prueba en 1, y repite.

    Selecciona la combinación que minimiza el MSE (Error Cuadrático Medio).

````python
from sklearn.model_selection import GridSearchCV

# Pipeline con parámetros variables
pipe = Pipeline([
    ('poly', PolynomialFeatures()),
    ('model', Ridge())
])

param_grid = {
    'poly__degree': [1, 2, 3, 4, 5,6,7],
    'model__alpha': [0.01, 0.1, 1.0, 5.0, 10.0]
}

search = GridSearchCV(pipe, param_grid, cv=3, scoring='neg_mean_squared_error')
search.fit(X, y)

print("Mejores parámetros:", search.best_params_)
print("Mejor score:", -search.best_score_)
````

````python
import matplotlib.pyplot as plt

# Modelo entrenado con mejores parámetros
best_model = search.best_estimator_
x_plot = np.linspace(min(x), max(x), 200).reshape(-1, 1)
y_plot = best_model.predict(x_plot)

plt.figure(figsize=(8,5))
plt.scatter(x, y, label="Datos", alpha=0.6)
plt.plot(x_plot, y_plot, color='red', label="Mejor ajuste")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Ajuste polinomial optimizado")
plt.legend()
plt.show()
````
