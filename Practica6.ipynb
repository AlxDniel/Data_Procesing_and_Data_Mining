## 🌀 1. K-Means Clustering

### 📖 Resumen
El algoritmo K-Means particiona datos en **K grupos** basados en la distancia a **centroides**.  
Requiere especificar previamente el número de grupos `K`.

**Pasos del algoritmo:**
1. Inicializar `K` centroides aleatoriamente.
2. Asignar cada punto al centro más cercano.
3. Recalcular los centroides.
4. Repetir hasta converger.


### 🧩 Ejemplo 1.1: K-Means con K en datos sintéticos

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

k = 5 # Numero de cluster a separar

# Generar datos sintéticos
X, y_true = make_blobs(n_samples=800, centers=k, cluster_std=0.90, random_state=1)

# Aplicar K-Means con K
kmeans = KMeans(n_clusters=k, random_state=10).fit(X)
y_kmeans = kmeans.predict(X)

# Visualizar
plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, cmap='viridis')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=300, c='red', marker='X')
plt.title(f'K-Means Clustering con K={k}')
plt.show()
```

```python
# Semilla para reproducibilidad
np.random.seed(0)

# Crear datos sintéticos
x1 = np.random.randint(0, 100, 800)
x2 = np.random.randint(0, 100, 800)

# Combinar en un array de forma (500, 2)
X = np.column_stack((x1, x2))

# Clustering con KMeans
for k in [2,3, 4, 6, 8]:
  kmeans = KMeans(n_clusters=k, random_state=1).fit(X)
  labels = kmeans.labels_

  # Graficar resultados
  plt.show()
  plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis')
  plt.title(f"Clustering con KMeans={k}")
  plt.xlabel("Feature 1")
  plt.ylabel("Feature 2")
  plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=300, c='red', marker='X')
  plt.legend()
  plt.show()
```

## 🌀 2. Hierarchical Clustering

### 📖 Resumen
El **clustering jerárquico** organiza los datos en una estructura en forma de árbol (dendrograma),  
sin necesidad de definir el número de clústeres previamente.

**Pasos del algoritmo:**
1. Considera cada punto como un clúster individual.
2. Fusiona los dos clústeres más cercanos repetidamente.
3. Representa las fusiones en un dendrograma.


### 🧩 Ejemplo 2.1: Clustering jerárquico en datos sintéticos con dendrograma

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage, fcluster
from sklearn.datasets import make_blobs

# Generar datos sintéticos
X, y_true = make_blobs(n_samples=50, centers=5, cluster_std=0.30, random_state=0)

# Calcular linkage matrix
Z = linkage(X, 'ward')

# Graficar dendrograma
plt.figure(figsize=(10, 5))
dendrogram(Z)
plt.title('Dendrograma - Clustering Jerárquico')
plt.show()

# Cortar el dendrograma en 3 clústeres
labels = fcluster(Z, t=3, criterion='maxclust')

# Visualizar partición
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis')
plt.title('Clustering Jerárquico - 3 Grupos')
plt.show()
```

```python
# Semilla para reproducibilidad
np.random.seed(0)

# Crear datos sintéticos
x1 = np.sin(np.random.randint(0, 100, 800))
x2 = np.cos(np.random.randint(0, 100, 800))

# Combinar en un array de forma (500, 2)
X = np.column_stack((x1, x2))

Z = linkage(X, 'ward')
# Graficar dendrograma
plt.figure(figsize=(10, 5))
dendrogram(Z)
plt.title('Dendrograma - Clustering Jerárquico')
plt.show()

# Cortar el dendrograma en 3 clústeres
labels = fcluster(Z, t=8, criterion='maxclust')

# Visualizar partición
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis')
plt.title('Clustering Jerárquico - 8 Grupos')
plt.show()
```

## 🌀 3. DBSCAN (Density-Based Spatial Clustering)

### 📖 Resumen
**DBSCAN** es un algoritmo de clustering basado en densidad que:
- Detecta clústeres de cualquier forma,
- Identifica **outliers** (puntos ruidosos),
- No requiere definir previamente el número de clústeres.

**Parámetros principales:**
- `eps`: radio de vecindad,
- `min_samples`: mínimo número de puntos para formar un clúster.



### 🧩 Ejemplo 3.1: DBSCAN en datos sintéticos con ruido

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN

# Generar datos en forma de media luna con ruido
X, _ = make_moons(n_samples=300, noise=0.15, random_state=0)

# Aplicar DBSCAN
db = DBSCAN(eps=0.2, min_samples=5).fit(X)
labels = db.labels_

# Visualizar (ruido etiquetado como -1 en DBSCAN)
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis')
plt.title('DBSCAN - Agrupamiento basado en densidad')
plt.show()
```

## 🌀 4. Gaussian Mixture Models (GMM)

### 📖 Resumen
**GMM** (Modelos de Mezcla Gaussiana) modela los datos como una combinación de varias distribuciones gaussianas.  
Asigna probabilísticamente cada punto a cada componente gaussiano.

**Ventajas:**
- Maneja bien clústeres con forma elíptica,
- Permite **probabilidades de pertenencia**.



### 🧩 Ejemplo 4.1: GMM en datos sintéticos con forma elíptica

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.mixture import GaussianMixture

# Generar datos sintéticos
X, y_true = make_blobs(n_samples=300, centers=3, cluster_std=0.60, random_state=0)

# Aplicar GMM
gmm = GaussianMixture(n_components=3, random_state=0).fit(X)
labels = gmm.predict(X)

# Visualizar
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis')
plt.title('Gaussian Mixture Model - 3 Componentes')
plt.show()
```

# 🧩 Ejemplo 4.2: Aplicación de GMM en Datos Unidimensionales

### 🎯 Objetivo:
Aplicar **Gaussian Mixture Models (GMM)** a un **histograma unidimensional**  
para identificar posibles **subpoblaciones** o **componentes estadísticas**  
en un conjunto de datos simulados, como colores o magnitudes.

---

## 🧭 Instrucciones:

1. Simule o cargue un conjunto de datos **unidimensionales**, por ejemplo:
   - Colores (BP-RP),
   - Magnitudes,
   - Velocidades radiales.
   
2. Construya un **histograma** de los datos con bins apropiados.

3. Aplique **GMM** con **2 o más componentes** directamente sobre los **valores individuales**,  
   no sobre los conteos del histograma.

4. Grafique el **histograma de los datos** junto a las **curvas de las componentes GMM**  
   y la **mezcla total ajustada**.

---

## ❓ Preguntas:

- ¿El GMM logra identificar **subpoblaciones** distintas?
- ¿Qué significa cada componente en el contexto físico de los datos?
- ¿Cómo cambia el resultado al variar el número de componentes?

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture

# Simular datos bimodales (ej: dos poblaciones estelares en color)
np.random.seed(42)
data1 = np.random.normal(0.75, 0.1, 500)   # Población 1
data2 = np.random.normal(0.99, 0.1, 300)   # Población 2
data = np.concatenate([data1, data2])

# Histograma de los datos
plt.hist(data, bins=30, density=True, alpha=0.6, color='gray')

# Aplicar GMM
gmm = GaussianMixture(n_components=2)
gmm.fit(data.reshape(-1, 1))

# Evaluar GMM sobre un rango continuo
x = np.linspace(data.min() - 0.2, data.max() + 0.2, 1000).reshape(-1, 1)
logprob = gmm.score_samples(x)
pdf = np.exp(logprob)

# Graficar GMM ajustado
plt.plot(x, pdf, '-r', label='GMM Total')

# Graficar cada componente individual
weights = gmm.weights_
means = gmm.means_.flatten()
covariances = gmm.covariances_.flatten()
for i in range(2):
    component_pdf = weights[i] * (1/np.sqrt(2 * np.pi * covariances[i])) * \
                    np.exp(-(x.flatten() - means[i])**2 / (2 * covariances[i]))
    plt.plot(x, component_pdf, '--', label=f'Componente {i+1}')

plt.xlabel('Valor observado (ej: BP-RP)')
plt.ylabel('Densidad')
plt.legend()
plt.title('Ajuste de GMM sobre datos unidimensionales')
plt.show()
```

## 🌀 5. Función de Correlación de Dos Puntos

### 📖 Resumen
La **función de correlación de dos puntos** mide el exceso de pares de puntos separados por una distancia `r` respecto a una distribución aleatoria.

**Estimador de Landy & Szalay (1993):**
\[
\hat{\xi}(r) = \frac{DD(r) - 2DR(r) + RR(r)}{RR(r)}
\]

Donde:
- `DD(r)`: número de pares de puntos reales,
- `RR(r)`: número de pares en una distribución aleatoria,
- `DR(r)`: número de pares cruzados entre datos y random.

---

## 🧩 Ejemplo 5.1: Función de correlación con datos simulados


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.spatial import distance_matrix

# Generar datos de puntos simulados (Data)
np.random.seed(0)
data = np.random.uniform(0, 1, (100, 2))

# Generar puntos aleatorios (Random)
random = np.random.uniform(0, 1, (100, 2))

# Calcular matrices de distancia
D_D = distance_matrix(data, data)
D_R = distance_matrix(data, random)
R_R = distance_matrix(random, random)

# Definir bins
bins = np.linspace(0, 0.5, 15)

# Contar pares en bins (excluir autopaired en DD y RR)
def count_pairs(dist_matrix, bins):
    pairs = []
    for i in range(len(bins)-1):
        mask = (dist_matrix >= bins[i]) & (dist_matrix < bins[i+1])
        count = np.sum(mask) - np.sum(np.diag(mask))  # Exclude diagonal if needed
        pairs.append(count)
    return np.array(pairs)

DD_counts = count_pairs(D_D, bins)
DR_counts = count_pairs(D_R, bins)
RR_counts = count_pairs(R_R, bins)

# Aplicar estimador de Landy-Szalay
xi = (DD_counts - 2*DR_counts + RR_counts) / RR_counts

# Graficar
bin_centers = 0.5 * (bins[1:] + bins[:-1])

plt.plot(bin_centers, xi, marker='o')
plt.xlabel('Separación (r)')
plt.ylabel('Función de Correlación ξ(r)')
plt.title('Función de Correlación de Dos Puntos - Datos Simulados')
plt.grid(True)
plt.show()
```

y visualizar tanto los **datos observados** como la **recta ajustada** en el mismo gráfico.

---

## 🧭 Instrucciones:

1. Simular datos con tendencia lineal y ruido.
2. Aplicar un ajuste de tipo **recta** usando `numpy.polyfit`.
3. Graficar los **datos simulados** junto a la **recta ajustada**.
4. Mostrar los **parámetros del ajuste** (pendiente y ordenada al origen).

```python
import numpy as np
import matplotlib.pyplot as plt

# Simular datos lineales con ruido
np.random.seed(42)
x_data = np.linspace(0, 10, 50)
y_data = 2.5 * x_data + 1.0 + np.random.normal(0, 2, size=len(x_data))

# Ajustar una recta
coeffs = np.polyfit(x_data, y_data, 1)  # Grado 1 -> lineal
slope, intercept = coeffs
print(f"Ajuste lineal: y = {slope:.2f} * x + {intercept:.2f}")

# Evaluar la recta ajustada
y_fit = slope * x_data + intercept

# Graficar
plt.figure(figsize=(8, 6))
plt.scatter(x_data, y_data, label='Datos simulados', color='black')
plt.plot(x_data, y_fit, label=f'Ajuste: y = {slope:.2f}x + {intercept:.2f}', color='red')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Ajuste Lineal sobre Datos Simulados')
plt.legend()
plt.grid(True)
plt.show()
```

# 🧩 Ejercicio – Función de Correlación de Dos Puntos en el Sloan Great Wall (SDSS)

## 🎯 Objetivo
Explorar la **distribución espacial de galaxias** dentro del **Sloan Great Wall**

## 🧭 Instrucciones:

1. Utilice el catálogo de galaxias que obtuvo de **SDSS** para la región del **Sloan Great Wall**.
2. Calcule las **distancias angulares o espaciales** entre todas las parejas de galaxias.
3. Construya un **histograma de conteo de pares** en función de la separación.
4. Normalice usando **pares aleatorios** para obtener la **función de correlación estimada**:

   \[
   \xi(r) = \frac{DD(r)}{RR(r)} - 1
   \]

   - \( DD(r) \): Conteo de pares en el catálogo real,
   - \( RR(r) \): Conteo de pares en un catálogo aleatorio con la misma geometría.

5. Grafique la función de correlación en escala log-log.

6. **Ajuste una ley de potencia** del tipo:

   \[
   $\xi(r) = \left( \frac{r}{r_0} \right)^{-\gamma}$
   \]

---

## ❓ Preguntas de Análisis:

- ¿Observa un **exceso de correlación** a alguna escala característica?
- ¿Qué valores obtiene para $( r_0 )$ y $( \gamma )$?
- ¿Cómo se comparan estas escalas con las típicas de **superestructuras cósmicas** (~5–10 Mpc)?
- ¿Qué implicaciones tiene esto sobre la **estructura a gran escala del universo**?

---

## ✅ Hint:
- Puede usar `scipy.spatial.distance.pdist` para calcular todas las distancias entre pares.
- Para el catálogo aleatorio, puede **simular posiciones uniformes** dentro del rango RA-Dec del Sloan Great Wall.



```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.spatial import distance
from scipy.optimize import curve_fit

# Simular un catálogo de galaxias en 2D (RA, Dec en grados)
np.random.seed(42)
n_galaxies = 500
ra = np.random.uniform(170, 210, n_galaxies)
dec = np.random.uniform(0, 10, n_galaxies)

# Convertir a radianes para cálculo de distancia angular
coords = np.deg2rad(np.column_stack((ra, dec)))

# Función para calcular distancia angular
def angular_distance(coord1, coord2):
    delta_ra = coord1[0] - coord2[0]
    delta_dec = coord1[1] - coord2[1]
    a = np.sin(delta_dec / 2)**2 + np.cos(coord1[1]) * np.cos(coord2[1]) * np.sin(delta_ra / 2)**2
    return 2 * np.arcsin(np.sqrt(a))

# Calcular todas las distancias reales (DD)
distances_dd = distance.pdist(coords, metric=lambda u, v: angular_distance(u, v))
distances_dd_deg = np.rad2deg(distances_dd)

# Generar catálogo aleatorio
ra_rand = np.random.uniform(170, 210, n_galaxies)
dec_rand = np.random.uniform(0, 10, n_galaxies)
coords_rand = np.deg2rad(np.column_stack((ra_rand, dec_rand)))

# Calcular todas las distancias aleatorias (RR)
distances_rr = distance.pdist(coords_rand, metric=lambda u, v: angular_distance(u, v))
distances_rr_deg = np.rad2deg(distances_rr)

# Crear histogramas
bins = np.logspace(-1, 1, 15)  # Bins en grados
dd_hist, _ = np.histogram(distances_dd_deg, bins=bins)
rr_hist, _ = np.histogram(distances_rr_deg, bins=bins)

# Calcular xi(r)
xi = dd_hist / rr_hist - 1

# Graficar función de correlación
r_centers = 0.5 * (bins[1:] + bins[:-1])

plt.figure(figsize=(8, 6))
plt.plot(r_centers, xi, 'o', label='Función de correlación estimada')
plt.xlabel('Separación angular [grados]')
plt.ylabel('$\\xi(\\theta)$')
plt.title('Función de Correlación de Dos Puntos - Sloan Great Wall')
plt.grid(True, which="both", ls="--", lw=0.5)

# Ajuste de ley de potencia
def power_law(r, r0, gamma):
    return (r / r0) ** (-gamma)

# Filtrar valores válidos para ajuste
valid = xi > 0
popt, _ = curve_fit(power_law, r_centers[valid], xi[valid], p0=[1.0, 1.0])

# Graficar ajuste
r_fit = np.logspace(-1, 1, 100)
xi_fit = power_law(r_fit, *popt)
plt.plot(r_fit, xi_fit, '-', label=f'Ajuste: $r_0 = {popt[0]:.2f}$, $\\gamma = {popt[1]:.2f}$', color='red')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
plt.plot(r_centers, xi, 'o', label='Función de correlación estimada')
plt.xlabel('Separación angular [grados]')
plt.ylabel('$\\xi(\\theta)$')
plt.title('Función de Correlación de Dos Puntos - Sloan Great Wall')
plt.grid(True, which="both", ls="--", lw=0.5)

# Ajuste de ley de potencia
def power_law(r, r0, gamma):
    return (r / r0) ** (-gamma)

# Filtrar valores válidos para ajuste
valid = xi > 0
popt, _ = curve_fit(power_law, r_centers[valid], xi[valid], p0=[1.0, 1.0])

# Graficar ajuste
r_fit = np.logspace(-1, 1, 100)
xi_fit = power_law(r_fit, *popt)
plt.plot(r_fit, xi_fit, '-', label=f'Ajuste: $r_0 = {popt[0]:.2f}$, $\\gamma = {popt[1]:.2f}$', color='red')
plt.legend()
plt.show()
```
