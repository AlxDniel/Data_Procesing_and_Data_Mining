# 🧩 Ejemplo Guiado – Clasificación de flores Iris con Redes Neuronales Artificiales (ANN)

## 🎯 Objetivo
Entrenar una red neuronal artificial simple para clasificar correctamente flores del conjunto de datos **Iris** en tres especies: *setosa*, *versicolor* y *virginica*, basándose en cuatro características morfológicas (largo y ancho de sépalos y pétalos).

El propósito de este ejercicio es entender cómo funciona una arquitectura ANN sencilla y evaluar su desempeño mediante métricas estándar de clasificación.

---

## 📋 Descripción del Dataset
El dataset **Iris** es uno de los conjuntos de datos más usados en aprendizaje automático. Cada muestra corresponde a una flor Iris, y tiene cuatro atributos numéricos:

- `sepal length (cm)`
- `sepal width (cm)`
- `petal length (cm)`
- `petal width (cm)`

Y una clase de salida:

- `species`: (*setosa*, *versicolor*, *virginica*)

## ✅ Pasos a desarrollar

1. Cargar y visualizar los datos para entender su distribución.
2. Preprocesar los datos: normalización y partición en train/test.
3. Construir una red neuronal densa (ANN) usando Keras:
   - Capas ocultas con activaciones no lineales (ReLU).
   - Capa de salida con softmax.
4. Entrenar el modelo y validar su desempeño.
5. Evaluar con:

   - Matriz de confusión
   - Reporte de clasificación
   - Curvas de pérdida y precisión

```python
# 🌸 Clasificación con ANN - Dataset Iris

# 1. Cargar librerías necesarias
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
```

```python

# 2. Cargar el dataset Iris
data = load_iris()
X = data.data
y = data.target

# 2.1 Visualizar el dataset
import pandas as pd
iris_df = pd.DataFrame(X, columns=data.feature_names)
iris_df['species'] = [data.target_names[i] for i in y]
sns.pairplot(iris_df, hue='species')
plt.suptitle("Visualización del Dataset Iris", y=1.02)
plt.show()

```

```python

# 3. Escalar los datos
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 4. Dividir en train y test
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# 5. Construir el modelo ANN
model = Sequential()
model.add(Dense(16, activation='relu', input_shape=(X.shape[1],)))
model.add(Dropout(0.1))
model.add(Dense(8, activation='relu'))
model.add(Dense(3, activation='softmax'))  # 3 clases

# 6. Compilar el modelo
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# 7. Entrenar el modelo
history = model.fit(X_train, y_train, epochs=100, batch_size=8, validation_split=0.1, verbose=0)

```

```python
# 8. Evaluar el modelo
loss, acc = model.evaluate(X_test, y_test, verbose=0)
print(f"\n\033[1mEvaluación en test set:\033[0m\nLoss: {loss:.4f} | Accuracy: {acc:.4f}")

# 9. Reporte de clasificación
y_pred = np.argmax(model.predict(X_test), axis=1)
print("\n\033[1mReporte de Clasificación:\033[0m")
print(classification_report(y_test, y_pred, target_names=data.target_names))

```

```python
# 10. Matriz de confusión
conf = confusion_matrix(y_test, y_pred)
sns.heatmap(conf, annot=True, fmt='d', cmap='Blues', xticklabels=data.target_names, yticklabels=data.target_names)
plt.xlabel("Predicted")
plt.ylabel("True")
plt.title("Confusion Matrix")
plt.show()

# 11. Curvas de entrenamiento
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.plot(history.history['accuracy'], label='Train Acc')
plt.plot(history.history['val_accuracy'], label='Val Acc')
plt.xlabel("Epoch")
plt.ylabel("Metric")
plt.title("Training History")
plt.legend()
plt.show()
```
