# 🧠 Evaluación Final – Proyecto de Redes Neuronales Aplicadas a Astronomía o Finanzas

## 🎯 Objetivo General

Desarrollar un proyecto final donde apliques lo aprendido sobre **redes neuronales** a un fenómeno o conjunto de datos de tu interés. El proyecto puede consistir en:

- El diseño de una red (propuesta argumentada y fundamentada), o
- La implementación de una red neuronal que clasifique, prediga o modele un fenómeno real.

Este trabajo puede entregarse en formato **notebook ejecutable (`.ipynb`)** o como **informe PDF estructurado**, siempre acompañado de código si hay implementación.

---

## 📋 Instrucciones Generales

Tu informe debe contener las siguientes secciones mínimas:

1. **Título del proyecto**  
2. **Motivación y contexto**: ¿Qué fenómeno deseas explorar o modelar? ¿Por qué es relevante?
3. **Descripción del conjunto de datos**: fuente, número de muestras, tipo de datos, preprocesamiento.
4. **Propuesta o implementación de la red neuronal**:
   - Tipo de red elegida (ANN, CNN, RNN, RecNN)
   - Arquitectura propuesta: capas, funciones de activación, optimizador, etc.
   - Justificación de las decisiones técnicas
5. **Entrenamiento del modelo y evaluación**:
   - Gráficos de pérdida y precisión
   - Métricas como matriz de confusión, curvas ROC (Curvas de Aprendisaje), RMSE, etc.
6. **Análisis y Resultados**:

   -  ¿Qué ventajas tiene esta arquitectura para tu problema a diferencia de otras?
   - ¿Qué desafíos o limitaciones tuvo tu modelo al implementarla?
   - ¿Qué aprendiste sobre los datos y la arquitectura?
   - ¿Cómo podrías mejorarlo?
   - ¿Qué métricas usaste para evaluar el rendimiento de tu modelo? ¿Por qué?
   - ¿Tu modelo está sobreajustado o subajustado? ¿Cómo lo sabes?
   - ¿Qué tan confiables son las predicciones o clasificaciones generadas?
   - Si el modelo no funcionó bien, ¿qué harías diferente?
   - ¿Qué mejorarías en una siguiente iteración?
   - ¿Tu red neuronal aprendió algo significativo del fenómeno?


7. **Conclusiones** y próximos pasos

## 🧠 Tipos de Redes Neuronales y sus Aplicaciones

| Tipo de Red | Aplicación típica | Fortalezas | Ejemplo |
|-------------|-------------------|------------|---------|
| **ANN** (Artificial Neural Network) | Datos tabulares, predicciones generales | Simples, flexibles | Precio de casas, calificación de riesgo |
| **CNN** (Convolutional Neural Network) | Imágenes, patrones espaciales | Captura estructuras visuales | Detección en imágenes satelitales o gráficas |
| **RNN / LSTM / GRU** | Series temporales, texto | Modela secuencias | Precio del dólar, series macroeconómicas |
| **RecNN** (Recursive Neural Network) | Estructuras jerárquicas o árboles | Procesa relaciones estructuradas | Análisis de documentos financieros por jerarquía |
